```python
def countChars(s):
    print(f'Len: {len(s)}')

countChars(p)
```

2. Contar el número de caracteres sin incluir los espacios

```python
def countCharsNoSpaces(s):
    chNS = s.replace(' ', '')
    print(f'Len (no spaces): {len(chNS)}')

countCharsNoSpaces(p)
```

3. Contar la cantidad de vocales que hay en el texto

```python
def countVowels(s):
    countV = 0
    for i in s:
        if i == 'a' or i == 'e' or i == 'i' or i == 'o' or i == 'u':
            countV += 1
    
    return countV

print(f'Total of Vowels: {countVowels(p)}')
```

## Resolución

Crea una agenda vacía

```python
contacts = {}
```

En la siguiente celda, incluye el código para ejecutar las siguientes instrucciones:
- Mostrar en pantalla las opciones
- Pedirle al usuario que elija las opciones
- Crear una estructura de control (*if*, *elif* y *else*) que ejecute diferentes bloques de código según las opciones elegidas.

```python
def adressBook():
    print(f'- - - C O N T A C T S - - -')
    print(f'Select an option to continue:')
    print(f'1. Add a contact.')
    print(f'2. Search a contact.')
    print(f'3. Edit a contact')
    print(f'4. Delete a contact')
    print(f'5. Show full adress book')
    print(f'6. Exit.')
    op = 0
    while op != 6:
        op = int(input(f'Introduce your option: '))
        if op == 1:
            name = input(f'Introduce your contact´s name: ')
            tel = int(input(f'Introduce your contact´s telephone: '))

            if name in contacts:
                print(f'This contact already exist')
            else:
                contacts[name] = tel
        elif op == 2:
            searchName = input(f'Search a contact by name, introduce it: ')
            key = contacts.get(searchName)

            if key is not None:
                print(f'Name: {searchName}')
                print(f'Number: {contacts[searchName]}')
            else:
                print(f'That contact does not exist')
        elif op == 3:
            contactToEdit = input(f'Introduce the contact you want to modify: ')

            if contactToEdit in contacts:
                n = int(input(f'Introduce the new number for {contactToEdit}'))
                contacts[contactToEdit] = n
                print(f'Change saved...')
            else:
                print(f'{contactToEdit} is not in your contact list')
        elif op == 4:
            contactToDelete = input(f'Introduce the name of the contact you want to delete: ')

            if contactToDelete in contacts:
                del contacts[contactToDelete]
                print(f'Contact deleted...')
            else:
                print(f'Contact did not found')
        elif op == 5:
            print(f'{contacts}')
        elif op == 6:
            print(f'Goodbay...')
            break
        else:
            print(f'Please enter a valid option')

adressBook()
```
